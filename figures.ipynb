```python
# =========================
# Figure 1
# =========================

import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams

# =========================
# 全局字体与出版级设置
config = {
    "font.family": 'serif',
    "font.size": 11,
    "mathtext.fontset": 'stix',
    "font.serif": ['Times New Roman', 'SimSun'],
    'axes.unicode_minus': False
}
rcParams.update(config)

# =========================
# 数据准备（来自 Table 1）
# 仅选取核心及显著差异指标，positive d = girls scored higher
indicators = [
    "Chinese Score***",
    "English Score**",
    "Backward Digit Span*",
    "WCST Categories (ns)",
    "AX-CPT d'-context (ns)",
    "Complex Subtraction (ns)"
]

cohen_d = np.array([0.46, 0.28, 0.21, 0.13, -0.10, -0.05])
p_values = ["< .001", ".001", ".015", ".140", ".279", ".531"]

y_pos = np.arange(len(indicators))

# =========================
# 创建图形
fig, ax = plt.subplots(figsize=(9, 4.5))

# 参考线 x = 0 (无差异线)
ax.axvline(0, linestyle="--", color="#7f7f7f", linewidth=1, zorder=1)

# 绘制点估计（区分显著与不显著：显著用实心，不显著用空心或灰色）
colors = ['#D95F02' if p in ["< .001", ".001", ".015"] else '#7f7f7f' for p in p_values]

for i in range(len(indicators)):
    ax.scatter(
        cohen_d[i],
        y_pos[i],
        color=colors[i],
        s=80,
        marker='o',
        zorder=3,
        edgecolors='black',
        linewidths=0.6
    )
    
    # 在右侧标注具体 Cohen's d 和 p 值
    text_label = f"d = {cohen_d[i]:+.2f}  (p {p_values[i]})"
    ax.text(
        0.60,
        y_pos[i],
        text_label,
        va='center',
        ha='left',
        fontsize=10,
        fontname='Times New Roman',
        color='#333333' if colors[i] != '#7f7f7f' else '#7f7f7f'
    )

# =========================
# 坐标轴与样式设置
ax.set_yticks(y_pos)
ax.set_yticklabels(indicators, fontsize=11, fontweight='bold', fontname='Times New Roman')
ax.set_xlim(-0.3, 0.8)
ax.set_ylim(-0.6, len(indicators) - 0.4)

ax.set_xlabel(
    "Standardized Sex Difference (Cohen's d: Girls − Boys)",
    fontsize=11,
    fontname='Times New Roman',
    labelpad=10
)

# 清理多余边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.tick_params(axis='y', length=0)
ax.tick_params(axis='x', labelsize=10)

# =========================
# 底部 Note 注释（融合 Table 1 背景）
#note = (
 #   "Note. N = 518 (Boys = 280, Girls = 238). Positive Cohen's d indicates higher mean scores for girls. "
 #   "Asterisks (*p < .05, **p < .01, ***p < .001) highlight significant sex differences, "
 #   "providing the empirical basis for freeing language intercepts in partial scalar invariance."
#)
#fig.text(
 #   0.5, 0.02, note, ha='center', va='bottom', fontsize=9, style='italic', color='#333333', fontname='Times New Roman'
#)

plt.tight_layout(rect=[0.02, 0.12, 0.98, 0.95])

# 保存高清图片
plt.savefig("figure2_sex_differences_effect_sizes.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
# =========================
# Figure 3
# =========================


import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams

# =========================
# 全局字体与样式设置 (APA 规范)
config = {
    "font.family": 'serif',
    "font.size": 11,
    "mathtext.fontset": 'stix',
    "font.serif": ['Times New Roman', 'SimSun'],
    'axes.unicode_minus': False
}
rcParams.update(config)

# =========================
# 数据准备
domains = ["Mathematics", "Chinese", "English"]

# 标准化间接效应 β
effects = np.array([0.232, 0.208, 0.206])

# 95% BCa CI
ci_lower = np.array([0.169, 0.146, 0.151])
ci_upper = np.array([0.301, 0.270, 0.268])

# 学术高级感配色（低饱和度、专业沉稳）
colors = ["#1f77b4", "#d62728", "#2ca02c"]  # 可更换为单色系或更柔和的学术色

# y轴位置（从上到下）
y_pos = np.array([3, 2, 1])

# =========================
# 创建图形 (更紧凑的宽高比)
fig, ax = plt.subplots(figsize=(9, 4.5))

# 参考线 x = 0 (使用虚线，颜色加深一点提升辨识度)
ax.axvline(0, linestyle="--", color="#7f7f7f", linewidth=1, zorder=1)

# =========================
# 绘制森林图
for i in range(len(domains)):
    # 95% 置信区间横线
    ax.hlines(
        y=y_pos[i],
        xmin=ci_lower[i],
        xmax=ci_upper[i],
        color=colors[i],
        linewidth=2,
        alpha=0.85,
        zorder=2
    )

    # 点估计（菱形）
    ax.scatter(
        effects[i],
        y_pos[i],
        color=colors[i],
        s=80,
        marker='D',
        zorder=3,
        edgecolors='black',
        linewidths=0.5
    )

    # 右侧统计量文本（采用规整的格式）
    text = f"$\\beta$ = {effects[i]:.3f}   95% BCa CI [{ci_lower[i]:.3f}, {ci_upper[i]:.3f}]"

    ax.text(
        0.34,  # 统一的文字起始横坐标
        y_pos[i],
        text,
        va='center',
        ha='left',
        fontsize=11,
        fontname='Times New Roman',
        zorder=3
    )

# =========================
# 坐标轴与布局设置
ax.set_yticks(y_pos)
ax.set_yticklabels(
    domains,
    fontsize=12,
    fontweight='bold',
    fontname='Times New Roman'
)

# 合理收紧 X 轴，避免右侧留白过多
ax.set_xlim(-0.05, 0.52)
ax.set_ylim(0.5, 3.5)

ax.set_xlabel(
    "Standardized Indirect Effect ($\\beta$)",
    fontsize=12,
    fontname='Times New Roman',
    labelpad=10
)

# 清理不必要的边框（保留底轴）
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# 隐藏 y 轴刻度线，保留标签
ax.tick_params(axis='y', length=0)
ax.tick_params(axis='x', labelsize=10, length=4)

for label in ax.get_xticklabels():
    label.set_fontname('Times New Roman')

# =========================
# 底部 Note 注释
#note = (
    "Note. Point estimates represent standardized indirect effects ($\\beta$). "
    "Error bars indicate 95% BCa bootstrap CIs (5,000 resamples).\n"
    "All indirect effects are statistically significant ($p < .001$)."
)

#fig.text(    0.5, 0.02,    note,    ha='center',    va='bottom',    fontsize=9.5,    style='italic',color='#333333',    fontname='Times New Roman',wrap=True)

# 自动布局调整
plt.tight_layout(rect=[0.02, 0.12, 0.98, 0.95])

# 保存出版级高清图片 (600 DPI TIFF/PNG)
plt.savefig(
    "figure4_forest_plot_refined.png",
    dpi=600,
    bbox_inches='tight'
)

plt.show()
```

```python
# =========================
# Figure 4
# =========================


import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams
from matplotlib.lines import Line2D

# =========================
# 全局字体与出版级设置
config = {
    "font.family": 'serif',
    "font.size": 10,
    "mathtext.fontset": 'stix',
    "font.serif": ['Times New Roman', 'SimSun'],
    'axes.unicode_minus': False
}
rcParams.update(config)

# =========================
# 创建画布
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(8.5, 3.0), gridspec_kw={'width_ratios': [1, 1]})

color_boys = '#2C7FB8'   # 沉稳蓝 (Boys)
color_girls = '#D95F02'  # 优雅橙红 (Girls)

# =========================
# 左侧子图 (Panel A)：Achievement Variance Explained (R²)
r2_boys = 72.7
r2_girls = 61.5
y_pos = np.array([0])
bar_height = 0.35

rects1 = ax1.barh(y_pos - bar_height/2, r2_boys, bar_height, color=color_boys, alpha=0.85, edgecolor='black', linewidth=0.7)
rects2 = ax1.barh(y_pos + bar_height/2, r2_girls, bar_height, color=color_girls, alpha=0.85, edgecolor='black', linewidth=0.7)

ax1.set_yticks(y_pos)
ax1.set_yticklabels(['Academic\nAchievement'], fontsize=10, fontweight='bold', fontname='Times New Roman')
ax1.set_xlim(0, 90)
ax1.set_xlabel("Variance Explained ($R^2$, %)", fontsize=10, fontname='Times New Roman', labelpad=6)
ax1.set_title("A. Variance Explained", fontsize=10.5, fontweight='bold', pad=8, fontname='Times New Roman')

# 标注数值
for rect in rects1:
    w = rect.get_width()
    ax1.annotate(f'{w:.1f}%', xy=(w, rect.get_y() + rect.get_height()/2),
                xytext=(5, 0), textcoords="offset points", ha='left', va='center', fontsize=9.5, fontname='Times New Roman')
for rect in rects2:
    w = rect.get_width()
    ax1.annotate(f'{w:.1f}%', xy=(w, rect.get_y() + rect.get_height()/2),
                xytext=(5, 0), textcoords="offset points", ha='left', va='center', fontsize=9.5, fontname='Times New Roman')

# =========================
# 右侧子图 (Panel B)：Standardized Indirect Effect (β)
# 极限缩短上下间距，去掉不专业的括号
beta_boys = 0.196
beta_girls = 0.288

# 纵向间距极度压缩至 0.02
ax2.scatter(beta_boys, 0.02, s=65, color=color_boys, marker='D', zorder=3, edgecolors='black', linewidths=0.5)
ax2.scatter(beta_girls, -0.02, s=65, color=color_girls, marker='o', zorder=3, edgecolors='black', linewidths=0.5)

# 绘制中心对齐虚线
ax2.axvline(0, linestyle="--", color="#7f7f7f", linewidth=0.7, zorder=1)

ax2.set_yticks([0])
ax2.set_yticklabels(['Indirect\nEffect'], fontsize=10, fontweight='bold', fontname='Times New Roman')
ax2.set_xlim(-0.05, 0.45)
ax2.set_ylim(-0.25, 0.25)  # 进一步收窄 Y 轴，极致紧凑
ax2.set_xlabel("Standardized Coefficient ($\\beta$)", fontsize=10, fontname='Times New Roman', labelpad=6)
ax2.set_title("B. Indirect Effects", fontsize=10.5, fontweight='bold', pad=8, fontname='Times New Roman')

# 纯净标注数值（无不专业括号）
ax2.annotate(f'$\\beta$ = {beta_boys:.3f}', xy=(beta_boys, 0.02), xytext=(6, -2), textcoords="offset points", ha='left', va='center', fontsize=9.5, fontname='Times New Roman')
ax2.annotate(f'$\\beta$ = {beta_girls:.3f}', xy=(beta_girls, -0.02), xytext=(6, -2), textcoords="offset points", ha='left', va='center', fontsize=9.5, fontname='Times New Roman')

# =========================
# 创建顶部共用图例 (Shared Legend)
legend_elements = [
    Line2D([0], [0], marker='s', color='w', label='Boys ($n$ = 280)', markerfacecolor=color_boys, markersize=8, markeredgecolor='black', markeredgewidth=0.5),
    Line2D([0], [0], marker='s', color='w', label='Girls ($n$ = 238)', markerfacecolor=color_girls, markersize=8, markeredgecolor='black', markeredgewidth=0.5)
]
fig.legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.50, 1.05), ncol=2, frameon=False, fontsize=9.5)

# =========================
# 样式精简与美化
for ax in [ax1, ax2]:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.tick_params(axis='y', length=0)
    ax.tick_params(axis='x', labelsize=9)

# =========================
# 底部 Note 注释
note = (
    r"Note. Structural paths are invariant across sex ($\Delta\chi^2(3) = 7.43, p = .059$). "
    r"The indirect effects do not differ significantly between groups ($p = .609$). "
    r"Differences in $R^2$ and $\beta$ reflect wider score dispersion among boys."
)
# fig.text(0.5, 0.01, note, ha='center', va='bottom', fontsize=8, style='italic', color='#333333', fontname='Times New Roman')

# 紧凑排版矩形范围
plt.tight_layout(rect=[0.02, 0.14, 0.98, 0.95])

# 保存高清出版级图片
plt.savefig("figure5_sex_invariance_pro.png", dpi=600, bbox_inches='tight')
plt.show()
```
